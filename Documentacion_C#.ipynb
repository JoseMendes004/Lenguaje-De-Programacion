# Introducción a C#

Este cuaderno presenta una introducción al lenguaje C#, su sintaxis y conceptos fundamentales para comenzar a programar.

## ¿Qué es C#?

C# (pronunciado *C Sharp*) es un lenguaje de programación moderno, orientado a objetos, desarrollado por Microsoft. Forma parte del ecosistema .NET y se utiliza para crear aplicaciones de escritorio, web, móviles, videojuegos, inteligencia artificial, entre otros.

```
// Ejemplo básico
using System;

class Program {
    static void Main() {
        Console.WriteLine("Hola, mundo");
    }
}
```

## Variables y Tipos de Datos

Una variable en C# debe declararse con un tipo de dato. Algunos tipos comunes son:

- `int`: números enteros
- `double`: números con decimales
- `char`: caracteres individuales
- `string`: cadenas de texto
- `bool`: valores lógicos (true o false)

También existen tipos derivados y personalizados mediante estructuras (`struct`) o clases (`class`).

```
int edad = 25;
double salario = 3500.75;
char inicial = 'J';
string nombre = "Ana";
bool esEmpleado = true;

Console.WriteLine($"Empleado: {nombre}, Edad: {edad}, Salario: {salario}");
```

## Operadores

C# incluye operadores para realizar cálculos y comparaciones.

**Aritméticos:** `+`, `-`, `*`, `/`, `%`

**Relacionales:** `==`, `!=`, `<`, `>`, `<=`, `>=`

**Lógicos:** `&&`, `||`, `!`

**Asignación:** `=`, `+=`, `-=`, `*=`, `/=`

```
int a = 10, b = 3;
Console.WriteLine(a + b);  // Suma
Console.WriteLine(a > b);  // Mayor que
Console.WriteLine(a != b); // Diferente de
a += 5;                    // a = a + 5
Console.WriteLine(a);
```

## Estructuras de Control

Las estructuras de control permiten ejecutar bloques de código dependiendo de condiciones o repeticiones.

```
int numero = 8;

if (numero % 2 == 0) {
    Console.WriteLine("El número es par.");
} else {
    Console.WriteLine("El número es impar.");
}

// Bucle for
for (int i = 0; i < 3; i++) {
    Console.WriteLine("Iteración " + (i + 1));
}

// Bucle while
int contador = 3;
while (contador > 0) {
    Console.WriteLine("Cuenta atrás: " + contador);
    contador--;
}
```

## Funciones o Métodos

Un método es un bloque de código que realiza una tarea específica. Puede o no devolver un valor y puede recibir parámetros.

```
static int Sumar(int x, int y) {
    return x + y;
}

Console.WriteLine(Sumar(5, 7)); // Llama al método y muestra 12
```

## Clases y Objetos

C# es un lenguaje orientado a objetos. Las clases definen las características y comportamientos de un tipo de objeto.

```
class Persona {
    public string Nombre;
    public int Edad;

    public void Saludar() {
        Console.WriteLine("Hola, mi nombre es " + Nombre + " y tengo " + Edad + " años.");
    }
}

Persona p1 = new Persona();
p1.Nombre = "Carlos";
p1.Edad = 30;
p1.Saludar();
```

## Propiedades y Encapsulamiento

En lugar de usar campos públicos, es una buena práctica usar **propiedades** para controlar el acceso a los datos internos de una clase.

```
class Estudiante {
    private string nombre;
    private double nota;

    public string Nombre {
        get { return nombre; }
        set { nombre = value; }
    }

    public double Nota {
        get { return nota; }
        set {
            if (value >= 0 && value <= 20) nota = value;
            else Console.WriteLine("Nota inválida.");
        }
    }
}

Estudiante e1 = new Estudiante();
e1.Nombre = "Lucía";
e1.Nota = 18;
Console.WriteLine($"Estudiante: {e1.Nombre}, Nota: {e1.Nota}");
```

## Arreglos

Un arreglo almacena varios valores del mismo tipo en una sola variable.

```
int[] numeros = {1, 2, 3, 4, 5};
foreach (int n in numeros) {
    Console.WriteLine(n);
}
```

## Listas

A diferencia de los arreglos, las listas son dinámicas y pueden cambiar de tamaño durante la ejecución. Requieren importar `System.Collections.Generic`.

```
using System.Collections.Generic;

List<string> frutas = new List<string>() {"Manzana", "Banana", "Cereza"};
frutas.Add("Pera");

foreach (string f in frutas) {
    Console.WriteLine(f);
}
```

## Manejo de Excepciones

Las excepciones permiten manejar errores en tiempo de ejecución y evitar que el programa se detenga abruptamente.

```
try {
    int x = 10;
    int y = 0;
    int resultado = x / y;
    Console.WriteLine(resultado);
} catch (DivideByZeroException e) {
    Console.WriteLine("Error: división entre cero.");
} finally {
    Console.WriteLine("Bloque finally ejecutado.");
}
```
