# Introducción a PHP

Este cuaderno presenta los conceptos fundamentales del lenguaje PHP, su sintaxis y ejemplos prácticos para quienes comienzan a programar en este lenguaje.

## ¿Qué es PHP?

PHP (Hypertext Preprocessor) es un lenguaje de programación de código abierto ampliamente utilizado en el desarrollo web. Permite crear páginas dinámicas e interactivas, integrándose fácilmente con HTML y bases de datos como MySQL.

```
<?php
// Primer programa en PHP
echo "Hola, mundo";
?>
```

## Variables y Tipos de Datos

En PHP, las variables comienzan con el signo `$`. No es necesario declarar el tipo de dato explícitamente.

Tipos básicos de datos en PHP:
- `int` → números enteros
- `float` → números con decimales
- `string` → texto
- `bool` → verdadero o falso
- `array` → colecciones de datos

```
<?php
$nombre = "Ana";
$edad = 25;
$altura = 1.68;
$es_estudiante = true;

echo "Nombre: $nombre, Edad: $edad, Altura: $altura m";
?>
```

## Operadores

PHP incluye varios tipos de operadores:

**Aritméticos:** `+`, `-`, `*`, `/`, `%`

**Relacionales:** `==`, `!=`, `>`, `<`, `>=`, `<=`

**Lógicos:** `&&`, `||`, `!`

**De asignación:** `=`, `+=`, `-=`, `*=`, `/=`

```
<?php
$a = 10;
$b = 5;

echo $a + $b; // Suma
echo $a > $b; // Comparación
$a += 2;
echo $a;      // Resultado: 12
?>
```

## Estructuras de Control

Las estructuras de control permiten ejecutar código condicional o repetitivo.

```
<?php
$numero = 7;

if ($numero % 2 == 0) {
    echo "El número es par";
} else {
    echo "El número es impar";
}

for ($i = 0; $i < 3; $i++) {
    echo "Iteración: $i<br>";
}

$contador = 3;
while ($contador > 0) {
    echo "Cuenta atrás: $contador<br>";
    $contador--;
}
?>
```

## Funciones

Una función en PHP se define con la palabra clave `function` y puede devolver un valor usando `return`.

```
<?php
function sumar($a, $b) {
    return $a + $b;
}

$resultado = sumar(3, 4);
echo "El resultado es: $resultado";
?>
```

## Arreglos

Un arreglo almacena múltiples valores bajo una sola variable. Puede ser **indexado** o **asociativo**.

```
<?php
// Arreglo indexado
$frutas = array("Manzana", "Banana", "Pera");
foreach ($frutas as $fruta) {
    echo $fruta . "<br>";
}

// Arreglo asociativo
$persona = array("nombre" => "Luis", "edad" => 30);
echo "Nombre: " . $persona["nombre"];
?>
```

## Manejo de Formularios

PHP se utiliza comúnmente para procesar formularios HTML. Los datos se reciben mediante los arreglos `$_GET` o `$_POST`.

```
<?php
// Ejemplo básico de procesamiento de formulario
if ($_SERVER["REQUEST_METHOD"] == "POST") {
    $nombre = $_POST["nombre"];
    echo "Hola, $nombre";
}
?>

<!-- Formulario HTML -->
<form method="post">
  <input type="text" name="nombre" placeholder="Escribe tu nombre">
  <input type="submit" value="Enviar">
</form>
```

## Manejo de Errores y Excepciones

PHP permite capturar errores usando `try`, `catch` y `throw`.

```
<?php
try {
    $divisor = 0;
    if ($divisor == 0) {
        throw new Exception("División entre cero");
    }
    echo 10 / $divisor;
} catch (Exception $e) {
    echo "Error: " . $e->getMessage();
}
?>
```
