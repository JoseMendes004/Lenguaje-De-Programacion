# Introducción a Ruby

Este cuaderno presenta los conceptos básicos del lenguaje Ruby, su sintaxis y ejemplos fundamentales para iniciarse en la programación con este lenguaje.

## ¿Qué es Ruby?

Ruby es un lenguaje de programación interpretado, orientado a objetos y de alto nivel. Es conocido por su sintaxis simple y legible, así como por su popular uso en el desarrollo web mediante el framework Ruby on Rails.

```
# Primer programa en Ruby
puts "Hola, mundo"
```

## Variables y Tipos de Datos

Ruby es un lenguaje **dinámicamente tipado**, lo que significa que no es necesario declarar el tipo de dato.

Tipos básicos:
- `Integer` → números enteros
- `Float` → números con decimales
- `String` → texto
- `Boolean` → `true` o `false`
- `Array` → lista ordenada de elementos
- `Hash` → pares clave-valor

```
nombre = "Ana"
edad = 25
altura = 1.68
es_estudiante = true

puts "Nombre: #{nombre}, Edad: #{edad}, Altura: #{altura}"
```

## Operadores

Ruby tiene operadores aritméticos, relacionales y lógicos.

**Aritméticos:** `+`, `-`, `*`, `/`, `%`

**Relacionales:** `==`, `!=`, `<`, `>`, `<=`, `>=`

**Lógicos:** `&&`, `||`, `!`

```
a = 10
b = 3

puts a + b   # Suma
puts a > b   # Mayor que
puts a != b  # Diferente
a += 2       # Incremento
puts a
```

## Estructuras de Control

Permiten controlar el flujo de ejecución del programa.

```
numero = 7

if numero.even?
  puts "El número es par"
else
  puts "El número es impar"
end

for i in 1..3
  puts "Iteración #{i}"
end

contador = 3
while contador > 0
  puts "Cuenta atrás: #{contador}"
  contador -= 1
end
```

## Métodos (Funciones)

Un método se define con la palabra clave `def`. Puede devolver un valor con `return`, aunque Ruby devuelve implícitamente el último valor evaluado.

```
def sumar(a, b)
  a + b
end

resultado = sumar(4, 5)
puts "El resultado es #{resultado}"
```

## Arreglos

Los arreglos almacenan múltiples valores en una sola variable. Pueden contener diferentes tipos de datos.

```
frutas = ["Manzana", "Banana", "Pera"]
frutas.each do |f|
  puts f
end
```

## Hashes

Un hash es una colección de pares clave-valor, similar a un diccionario en otros lenguajes.

```
persona = { "nombre" => "Carlos", "edad" => 30 }
puts "Nombre: #{persona["nombre"]}, Edad: #{persona["edad"]}"
```

## Clases y Objetos

Ruby es completamente orientado a objetos. Todo en Ruby es un objeto.

```
class Persona
  attr_accessor :nombre, :edad

  def initialize(nombre, edad)
    @nombre = nombre
    @edad = edad
  end

  def saludar
    puts "Hola, mi nombre es #{@nombre} y tengo #{@edad} años."
  end
end

p1 = Persona.new("Lucía", 22)
p1.saludar
```

## Manejo de Excepciones

Ruby permite manejar errores en tiempo de ejecución con `begin`, `rescue` y `ensure`.

```
begin
  x = 10
  y = 0
  resultado = x / y
rescue ZeroDivisionError => e
  puts "Error: división entre cero"
ensure
  puts "Bloque ensure ejecutado"
end
```
